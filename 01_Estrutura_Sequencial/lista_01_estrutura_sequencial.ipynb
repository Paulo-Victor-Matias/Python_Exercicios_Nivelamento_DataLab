## 📚 Revisão rápida

| Conceito | Exemplo |
|---|---|
| Ler um texto | `nome = input('Seu nome: ')` |
| Ler um número inteiro | `idade = int(input('Idade: '))` |
| Ler um número decimal | `altura = float(input('Altura: '))` |
| Operadores | `+  -  *  /` (divisão) · `//` (divisão inteira) · `%` (resto) · `**` (potência) |
| Mostrar com f-string | `print(f'Olá, {nome}!')` |
| Formatar casas decimais | `print(f'{valor:.2f}')` → `3.14` |
| Arredondar para cima | `import math` e depois `math.ceil(2.1)` → `3` |

> ⚠️ `input()` **sempre** devolve texto (`str`). Para fazer contas, converta com `int()` ou `float()`.

## Exercício 01 🟢 — Conversor de medidas

Faça um programa que peça uma medida em **metros** e mostre o valor equivalente em **centímetros** e em **milímetros**.

**Exemplo de execução:**

```text
Digite a medida em metros: 1.75
1.75 m equivalem a 175.0 cm e 1750.0 mm.
```

<details>
<summary>💡 <b>Dica</b> (clique para abrir)</summary>

1 metro = 100 centímetros = 1000 milímetros.

</details>

````python
# Exercício 01 — Conversor de medidas
# Escreva sua solução abaixo:
medida = float(input('Digite a medida em metros: '))
centimetros = medida * 100
milimetros = medida * 1000
print(f"{medida} m equivalem a {centimetros} cm e {milimetros} mm")
````

## Exercício 02 🟢 — Média bimestral

Faça um programa que peça as **4 notas bimestrais** de um aluno e mostre a **média** com duas casas decimais.

**Exemplo de execução:**

```text
Nota do 1º bimestre: 7
Nota do 2º bimestre: 8
Nota do 3º bimestre: 6.5
Nota do 4º bimestre: 9.5
Média final: 7.75
```

<details>
<summary>💡 <b>Dica</b> (clique para abrir)</summary>

Lembre-se dos parênteses: `(n1 + n2 + n3 + n4) / 4` é diferente de `n1 + n2 + n3 + n4 / 4`.

</details>

````python
# Exercício 02 — Média bimestral
# Escreva sua solução abaixo:
nota1 = float(input('Insira a nota do 1° bimestre: '))
nota2 = float(input('Insira a nota do 2° bimestre: '))
nota3 = float(input('Insira a nota do 3° bimestre: '))
nota4 = float(input('Insira a nota do 4° bimestre: '))
media = (nota1 + nota2 + nota3 + nota4) / 4
print(f"A média das notas é: {media}")
````
